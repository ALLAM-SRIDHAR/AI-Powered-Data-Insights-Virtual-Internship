```python
# Step 1: Clean and extract valid date formats
cleaned_end_dates = []

for date_str in df['Opportunity End Date']:
    # Check if the value is not NaN
    if pd.notna(date_str):
        # Split on space and take the first part (date portion only)
        date_part = date_str.split()[0]

        # Replace hyphens with slashes for consistency
        date_part = date_part.replace('-', '/')

        # Convert to datetime, extracting only the date part
        try:
            date = pd.to_datetime(date_part, errors='coerce')
            cleaned_end_dates.append(date.date() if pd.notna(date) else None)
        except Exception:
            cleaned_end_dates.append(None)
    else:
        cleaned_end_dates.append(None)

# Step 2: Assign the cleaned dates back to the 'Opportunity End Date' column
df['Opportunity End Date'] = cleaned_end_dates

# Step 3: Verify the changes
print(df.head())

# Optional: Check if there are any remaining missing values
missing_count = df['Opportunity End Date'].isna().sum()
print(f"Missing values after cleaning: {missing_count}")
```

```python

# Convert date columns to datetime
date_columns = ['Opportunity Start Date', 'Opportunity End Date', 'Reward Awarded Date', 'Completion Date']
for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Handle Opportunity Start Date
# Calculate the median duration between start and end dates for non-null pairs
valid_duration = df['Opportunity End Date'] - df['Opportunity Start Date']
median_duration = valid_duration.dropna().median()

# Fill missing start dates by subtracting the median duration from the end date
df.loc[df['Opportunity Start Date'].isnull(), 'Opportunity Start Date'] = df['Opportunity End Date'] - median_duration
```

```python
# Find rows with missing values in 'Institution Name' and 'Current/Intended Major'
missing_institution = df[df['Institution Name'].isna()][['Profile Id', 'Institution Name', 'Country', 'State', 'City', 'Address Line 1', 'Address Line 2']]
missing_major = df[df['Current/Intended Major'].isna()][['Profile Id', 'Current/Intended Major', 'Country', 'State', 'City', 'Address Line 1', 'Address Line 2']]

# Display the results
print("Missing Institution Name:")
print(missing_institution)

print("\nMissing Current/Intended Major:")
print(missing_major)
```

```python
# Get the most frequent 'Institution Name' for each city
institution_mode = df.groupby('City')['Institution Name'].agg(lambda x: x.mode()[0] if not x.mode().empty else 'Unknown Institution')

# Replace missing values in 'Institution Name' with the mode
for index in df[df['Institution Name'].isna()].index:
    city = df.loc[index, 'City']
    df.loc[index, 'Institution Name'] = institution_mode.get(city, 'Unknown Institution')
```

```python
# Get the overall most frequent value for 'Current/Intended Major'
major_mode = df['Current/Intended Major'].mode()[0] if not df['Current/Intended Major'].mode().empty else 'Undecided'

# Replace missing values in 'Current/Intended Major'
df['Current/Intended Major'].fillna(major_mode, inplace=True)
```

```python
# Define a function to convert the date formats
def convert_date_format(date):
    for fmt in ("%m/%d/%Y", "%d-%m-%Y"):  # Add more formats if needed
        try:
            return pd.to_datetime(date, format=fmt)
        except ValueError:
            continue
    return pd.NaT  # Return NaT if no format matches

# Apply the function to the 'Date of Birth' column
df['Date of Birth'] = df['Date of Birth'].apply(convert_date_format)

# Extract only the date part
df['Date of Birth'] = df['Date of Birth'].dt.date
```

```python
import re
# Function to process the 'Zipcode' column
def process_zipcode(row):
    zipcode = str(row['Zip Code']).strip()
    
    # Case 1: Format 60614-4904 -> Take first part before '-'
    if '-' in zipcode:
        return zipcode.split('-')[0]
    
    # Case 2: Format "485 garki" -> Extract numeric part only
    elif re.match(r'^\d+\s*\D+', zipcode):
        return re.findall(r'^\d+', zipcode)[0]
    
    # Case 3: Format like "Saifali1163", "WA189932", "Po box 10410", "MO 63108"
    elif re.search(r'\d+', zipcode):
        return re.findall(r'\d+', zipcode)[-1]
    
    # Case 4: If spaces between numeric values, e.g., 625 020 -> Remove spaces
    elif re.match(r'^\d+\s+\d+$', zipcode):
        return zipcode.replace(' ', '')
    
    # Default: Leave as-is
    return zipcode

# Apply the function to the 'Zipcode' column
df['Zip Code'] = df.apply(process_zipcode, axis=1)

# Display the updated DataFrame
print(df)
```
