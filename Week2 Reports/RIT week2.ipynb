```python
import seaborn as sns
import matplotlib.pyplot as plt

# List of categorical columns to plot
categorical_cols = [
    'Profile Id', 'Opportunity Name', 'Opportunity Category', 'Gender', 
    'Country', 'State', 'City', 'Institution Name', 'Current Student Status', 
    'Current/Intended Major', 'Badge Name', 'Skills Earned', 'Full_Name', 'Zip Code','Full_Address'
]

# Loop through each categorical column and plot top 10 categories
for col in categorical_cols:
    plt.figure(figsize=(10, 5))  # Adjust the figure size
    top_10_values = df[col].value_counts().head(10).index  # Get top 10 values
    sns.countplot(data=df, y=col, order=top_10_values)  # Plot only top 10
    plt.title(f'Top 10 Count Plot of {col}')
    plt.xlabel('Count')
    plt.ylabel(col)
    plt.show()
```

```python
# Step 1: Get the top 10 Current/Intended Majors
top_majors = df['Current/Intended Major'].value_counts().nlargest(10).index

# Step 2: Filter the DataFrame for top 10 majors
filtered_df = df[df['Current/Intended Major'].isin(top_majors)]

# Step 3: Create the count plot
plt.figure(figsize=(12, 8))
bar_plot = sns.countplot(data=filtered_df, y='Current/Intended Major', hue='Current Student Status', palette='pastel')

# Adding values on bars
for p in bar_plot.patches:
    bar_plot.annotate(f'{int(p.get_width())}', 
                      (p.get_width() + 1, p.get_y() + p.get_height() / 2), 
                      ha='left', va='center', fontsize=8, color='black')

# Wrapping long y-axis labels
def wrap_labels(labels, max_length=21):
    wrapped_labels = []
    for label in labels:
        if len(label) > max_length:
            # Wrap the label at the specified length
            wrapped_label = '\n'.join([label[i:i + max_length] for i in range(0, len(label), max_length)])
            wrapped_labels.append(wrapped_label)
        else:
            wrapped_labels.append(label)
    return wrapped_labels

# Get wrapped y-axis labels
wrapped_labels = wrap_labels(filtered_df['Current/Intended Major'].unique())
bar_plot.set_yticklabels(wrapped_labels)

# Title and labels
plt.title('Current Student Status by Current/Intended Major (Top 10)')
plt.xlabel('Count')
plt.ylabel('Current/Intended Major')
plt.legend(title='Current Student Status')

plt.tight_layout()  # Adjust layout
plt.show()
```
