```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns

# Import multiple algorithms
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

class StudentDropoutPredictor:
    def __init__(self, df):
        self.df = df
        self.models = {
            'Logistic Regression': LogisticRegression(random_state=42),
            'Decision Tree': DecisionTreeClassifier(random_state=42),
            'Random Forest': RandomForestClassifier(random_state=42),
            'Gradient Boosting': GradientBoostingClassifier(random_state=42),
            'SVM': SVC(random_state=42, probability=True),
            'KNN': KNeighborsClassifier(),
            'XGBoost': XGBClassifier(random_state=42)
        }
        self.model_scores = {}
        
    def prepare_data(self):
        """Prepare data for modeling"""
        # Create target variable
        self.df['dropout'] = self.df['Status Description'].map(
            lambda x: 1 if x in ['Dropped Out', 'Withdraw'] else 0
        )
        
        # Convert dates and create features
        date_columns = ['Learner SignUp DateTime', 'Opportunity End Date', 
                       'Date of Birth', 'Graduation Date', 'Entry created at', 
                       'Apply Date', 'Opportunity Start Date']
        
        for col in date_columns:
            self.df[col] = pd.to_datetime(self.df[col], format='%d-%m-%Y', errors='coerce')
        
        # Create derived features
        self.df['days_to_start'] = (self.df['Opportunity Start Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_end'] = (self.df['Opportunity End Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_graduate'] = (self.df['Graduation Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['application_processing_time'] = (self.df['Entry created at'] - self.df['Apply Date']).dt.days
        
        # Handle categorical variables
        categorical_columns = ['Gender', 'Country', 'Current Student Status', 
                             'Current/Intended Major', 'Opportunity Category']
        
        for col in categorical_columns:
            le = LabelEncoder()
            self.df[col + '_encoded'] = le.fit_transform(self.df[col])
        
        # Select features for modeling
        self.feature_columns = [
            'Age', 'days_to_start', 'days_to_end', 'days_to_graduate',
            'application_processing_time', 'Gender_encoded', 'Country_encoded',
            'Current Student Status_encoded', 'Current/Intended Major_encoded',
            'Opportunity Category_encoded'
        ]
        
        # Prepare X and y
        X = self.df[self.feature_columns]
        y = self.df['dropout']
        
        # Scale features
        scaler = StandardScaler()
        X_scaled = scaler.fit_transform(X)
        X_scaled = pd.DataFrame(X_scaled, columns=self.feature_columns)
        
        # Split data
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            X_scaled, y, test_size=0.2, random_state=42, stratify=y
        )
        
        print("Data preparation completed:")
        print(f"Training set shape: {self.X_train.shape}")
        print(f"Testing set shape: {self.X_test.shape}")
        print(f"Dropout rate in training set: {self.y_train.mean():.2%}")
        
    def train_and_evaluate_models(self):
        """Train and evaluate all models"""
        results = []
        
        for name, model in self.models.items():
            print(f"\nTraining {name}...")
            
            # Train model
            model.fit(self.X_train, self.y_train)
            
            # Make predictions
            y_pred = model.predict(self.X_test)
            y_pred_proba = model.predict_proba(self.X_test)[:, 1]
            
            # Calculate metrics
            accuracy = model.score(self.X_test, self.y_test)
            auc_score = roc_auc_score(self.y_test, y_pred_proba)
            
            # Cross-validation score
            cv_scores = cross_val_score(model, self.X_train, self.y_train, cv=5)
            
            results.append({
                'Model': name,
                'Accuracy': accuracy,
                'AUC-ROC': auc_score,
                'CV Mean': cv_scores.mean(),
                'CV Std': cv_scores.std()
            })
            
            # Print detailed report
            print(f"\nClassification Report for {name}:")
            print(classification_report(self.y_test, y_pred))
            
            # Store model scores
            self.model_scores[name] = {
                'accuracy': accuracy,
                'auc': auc_score,
                'cv_mean': cv_scores.mean(),
                'cv_std': cv_scores.std()
            }
        
        # Create comparison DataFrame
        self.results_df = pd.DataFrame(results)
        self.results_df = self.results_df.sort_values('AUC-ROC', ascending=False)
        
        return self.results_df
    
    def plot_model_comparison(self):
        """Plot model comparison results"""
        # Plot 1: Model Performance Comparison
        plt.figure(figsize=(12, 6))
        x = np.arange(len(self.results_df))
        width = 0.35
        
        plt.bar(x - width/2, self.results_df['Accuracy'], width, label='Accuracy')
        plt.bar(x + width/2, self.results_df['AUC-ROC'], width, label='AUC-ROC')
        
        plt.xlabel('Models')
        plt.ylabel('Score')
        plt.title('Model Performance Comparison')
        plt.xticks(x, self.results_df['Model'], rotation=45)
        plt.legend()
        plt.tight_layout()
        plt.show()
        
        # Plot 2: Cross-validation Results
        plt.figure(figsize=(10, 6))
        plt.errorbar(self.results_df['Model'], self.results_df['CV Mean'], 
                    yerr=self.results_df['CV Std'], fmt='o')
        plt.xlabel('Models')
        plt.ylabel('Cross-validation Score')
        plt.title('Cross-validation Results with Standard Deviation')
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()

# Execute the analysis
def run_analysis(df):
    print("Starting comprehensive model comparison...")
    
    # Initialize predictor
    predictor = StudentDropoutPredictor(df)
    
    # Prepare data
    predictor.prepare_data()
    
    # Train and evaluate models
    results = predictor.train_and_evaluate_models()
    
    # Plot comparisons
    predictor.plot_model_comparison()
    
    # Print final results
    print("\nFinal Model Comparison:")
    print(results.to_string(index=False))
    
    # Identify best model
    best_model = results.iloc[0]
    print(f"\nBest performing model: {best_model['Model']}")
    print(f"Accuracy: {best_model['Accuracy']:.4f}")
    print(f"AUC-ROC: {best_model['AUC-ROC']:.4f}")
    print(f"Cross-validation Score: {best_model['CV Mean']:.4f} (±{best_model['CV Std']:.4f})")
    
    return predictor

# Run the analysis (assuming df is your DataFrame)
predictor = run_analysis(df)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings
warnings.filterwarnings('ignore')

# Import models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

class AdvancedStudentDropoutPredictor:
    def __init__(self, df):
        self.df = df
        self.models = {
            'Logistic Regression': {
                'model': LogisticRegression(random_state=42),
                'params': {
                    'C': [0.001, 0.01, 0.1, 1, 10],
                    'penalty': ['l1', 'l2'],
                    'solver': ['liblinear']
                }
            },
            'Decision Tree': {
                'model': DecisionTreeClassifier(random_state=42),
                'params': {
                    'max_depth': [None, 10, 20, 30],
                    'min_samples_split': [2, 5, 10]
                }
            },
            'Random Forest': {
                'model': RandomForestClassifier(random_state=42),
                'params': {
                    'n_estimators': [100, 200],
                    'max_depth': [10, 20, None],
                    'min_samples_split': [2, 5]
                }
            },
            'Gradient Boosting': {
                'model': GradientBoostingClassifier(random_state=42),
                'params': {
                    'n_estimators': [100, 200],
                    'learning_rate': [0.01, 0.1],
                    'max_depth': [3, 5, 7]
                }
            },
            'Support Vector Machine': {
                'model': SVC(probability=True, random_state=42),
                'params': {
                    'C': [0.1, 1, 10],
                    'kernel': ['linear', 'rbf']
                }
            },
            'K-Neighbors Classifier': {
                'model': KNeighborsClassifier(),
                'params': {
                    'n_neighbors': [3, 5, 7, 9],
                    'weights': ['uniform', 'distance']
                }
            },
            'XGBoost': {
                'model': XGBClassifier(random_state=42),
                'params': {
                    'n_estimators': [100, 200],
                    'max_depth': [3, 5, 7],
                    'learning_rate': [0.01, 0.1]
                }
            }
        }
        self.best_models = {}
        self.feature_importances = {}
        self.scaler = StandardScaler()
        self.encoders = {}
        
    def prepare_data(self):
        """Prepare data with enhanced feature engineering"""
        print("Starting data preparation...")
        
        # Create target variable
        self.df['dropout'] = self.df['Status Description'].map(
            lambda x: 1 if x in ['Dropped Out', 'Withdraw'] else 0
        )
        
        # Convert dates and create features
        date_columns = ['Learner SignUp DateTime', 'Opportunity End Date', 
                        'Date of Birth', 'Graduation Date', 'Entry created at', 
                        'Apply Date', 'Opportunity Start Date']
        
        for col in date_columns:
            self.df[col] = pd.to_datetime(self.df[col], format='%d-%m-%Y', errors='coerce')
        
        # Enhanced feature engineering
        self.df['days_to_start'] = (self.df['Opportunity Start Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_end'] = (self.df['Opportunity End Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_graduate'] = (self.df['Graduation Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['application_processing_time'] = (self.df['Entry created at'] - self.df['Apply Date']).dt.days
        self.df['program_duration'] = (self.df['Opportunity End Date'] - self.df['Opportunity Start Date']).dt.days
        
        # Handle categorical variables
        categorical_columns = ['Gender', 'Country', 'Current Student Status', 
                               'Current/Intended Major', 'Opportunity Category']
        
        for col in categorical_columns:
            self.encoders[col] = LabelEncoder()
            self.df[col + '_encoded'] = self.encoders[col].fit_transform(self.df[col])
        
        # Select features
        self.feature_columns = [
            'Age', 'days_to_start', 'days_to_end', 'days_to_graduate',
            'application_processing_time', 'program_duration',
            'Gender_encoded', 'Country_encoded', 'Current Student Status_encoded',
            'Current/Intended Major_encoded', 'Opportunity Category_encoded'
        ]
        
        # Prepare X and y
        X = self.df[self.feature_columns]
        y = self.df['dropout']
        
        # Scale features
        X_scaled = self.scaler.fit_transform(X)
        X_scaled = pd.DataFrame(X_scaled, columns=self.feature_columns)
        
        # Split data with stratification
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            X_scaled, y, test_size=0.2, random_state=42, stratify=y
        )
        
        print("\nData preparation completed:")
        print(f"Training set shape: {self.X_train.shape}")
        print(f"Testing set shape: {self.X_test.shape}")
        print(f"Dropout rate: {self.y_train.mean():.2%}")
        
    def train_and_evaluate_models(self):
        """Train models with hyperparameter tuning and comprehensive evaluation"""
        results = []
        
        for name, model_info in self.models.items():
            print(f"\nTraining {name} with hyperparameter tuning...")
            
            # Perform grid search
            grid_search = GridSearchCV(
                model_info['model'],
                model_info['params'],
                cv=5,
                scoring='roc_auc',
                n_jobs=-1
            )
            grid_search.fit(self.X_train, self.y_train)
            
            # Store best model
            self.best_models[name] = grid_search.best_estimator_
            
            # Make predictions
            y_pred = grid_search.predict(self.X_test)
            y_pred_proba = grid_search.predict_proba(self.X_test)[:, 1]
            
            # Calculate metrics
            accuracy = grid_search.score(self.X_test, self.y_test)
            auc_score = roc_auc_score(self.y_test, y_pred_proba)
            cv_scores = cross_val_score(grid_search.best_estimator_, 
                                         self.X_train, self.y_train, cv=5)
            
            # Store results
            results.append({
                'Model': name,
                'Accuracy': accuracy,
                'AUC-ROC': auc_score,
                'CV Mean': cv_scores.mean(),
                'CV Std': cv_scores.std(),
                'Best Parameters': grid_search.best_params_
            })
            
            # Print detailed report
            print(f"\nBest parameters for {name}:")
            print(grid_search.best_params_)
            print(f"\nClassification Report for {name}:")
            print(classification_report(self.y_test, y_pred))
            
            # Calculate and plot confusion matrix
            self.plot_confusion_matrix(y_pred, name)
            
            # Calculate ROC curve
            self.plot_roc_curve(y_pred_proba, name)
            
            # Calculate feature importance
            if hasattr(grid_search.best_estimator_, 'feature_importances_'):
                self.feature_importances[name] = pd.DataFrame({
                    'feature': self.feature_columns,
                    'importance': grid_search.best_estimator_.feature_importances_
                }).sort_values('importance', ascending=False)
            elif name == 'Logistic Regression':
                self.feature_importances[name] = pd.DataFrame({
                    'feature': self.feature_columns,
                    'importance': abs(grid_search.best_estimator_.coef_[0])
                }).sort_values('importance', ascending=False)

        self.results_df = pd.DataFrame(results)
        self.plot_model_comparison()
        return self.results_df

    def plot_confusion_matrix(self, y_pred, model_name):
        """Plot confusion matrix for the model"""
        cm = confusion_matrix(self.y_test, y_pred)

        # Print the confusion matrix
        print(f"\nConfusion Matrix for {model_name}:\n")
        print(pd.DataFrame(cm, columns=['Predicted: Not Dropped Out', 'Predicted: Dropped Out'],
                           index=['True: Not Dropped Out', 'True: Dropped Out']))

        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                    xticklabels=['Not Dropped Out', 'Dropped Out'],
                    yticklabels=['Not Dropped Out', 'Dropped Out'])
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.title(f'Confusion Matrix for {model_name}')
        plt.show()

    def plot_roc_curve(self, y_pred_proba, model_name):
        """Plot ROC curve for the model"""
        fpr, tpr, _ = roc_curve(self.y_test, y_pred_proba)
        plt.figure(figsize=(8, 6))
        plt.plot(fpr, tpr, label='ROC curve', color='blue')
        plt.plot([0, 1], [0, 1], color='red', linestyle='--')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'ROC Curve for {model_name}')
        plt.legend()
        plt.show()

    def plot_model_comparison(self):
        """Visualize comparison of model performance metrics"""
        plt.figure(figsize=(12, 8))
        sns.barplot(data=self.results_df, x='Model', y='AUC-ROC', palette='viridis')
        plt.title('Model Comparison: AUC-ROC')
        plt.xticks(rotation=45)
        plt.show()

    def save_best_model(self, model_name):
        """Save the best Random Forest model to disk"""
        if model_name in self.best_models:
            joblib.dump(self.best_models[model_name], f'{model_name.replace(" ", "_")}_model.pkl')
            print(f"Saved {model_name} model to disk.")
        else:
            print(f"{model_name} not found in the best models.")

    def feature_importance_summary(self):
        for name, importance_df in self.feature_importances.items():
            plt.figure(figsize=(10, 6))
            sns.barplot(data=importance_df.head(10), x='importance', y='feature', palette='viridis')
            # Add values on bars
            for index, value in enumerate(importance_df.head(10)['importance']):
                plt.text(value, index, f'{value:.2f}', va='center')  # Adjust the format as needed
            plt.title(f'Feature Importance for {name}')
            plt.xlabel('Importance')
            plt.ylabel('Feature')
            plt.show()

# Example of using the class with a dataframe
# df = pd.read_csv('path_to_your_data.csv')
# predictor = AdvancedStudentDropoutPredictor(df)
# predictor.prepare_data()
# results = predictor.train_and_evaluate_models()
# print(results)
# predictor.save_best_model('Random Forest')
# predictor.feature_importance_summary()
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, roc_auc_score
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import warnings
warnings.filterwarnings('ignore')

# Import models
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

class AdvancedStudentDropoutPredictor:
    def __init__(self, df):
        self.df = df
        self.models = {
            'Logistic Regression': {
                'model': LogisticRegression(random_state=42),
                'params': {
                    'C': [0.001, 0.01, 0.1, 1, 10],
                    'penalty': ['l1', 'l2'],
                    'solver': ['liblinear']
                }
            },
            'Decision Tree': {
                'model': DecisionTreeClassifier(random_state=42),
                'params': {
                    'max_depth': [None, 10, 20, 30],
                    'min_samples_split': [2, 5, 10]
                }
            },
            'Random Forest': {
                'model': RandomForestClassifier(random_state=42),
                'params': {
                    'n_estimators': [100, 200],
                    'max_depth': [10, 20, None],
                    'min_samples_split': [2, 5]
                }
            },
            'Gradient Boosting': {
                'model': GradientBoostingClassifier(random_state=42),
                'params': {
                    'n_estimators': [100, 200],
                    'learning_rate': [0.01, 0.1],
                    'max_depth': [3, 5, 7]
                }
            },
            'Support Vector Machine': {
                'model': SVC(probability=True, random_state=42),
                'params': {
                    'C': [0.1, 1, 10],
                    'kernel': ['linear', 'rbf']
                }
            },
            'K-Neighbors Classifier': {
                'model': KNeighborsClassifier(),
                'params': {
                    'n_neighbors': [3, 5, 7, 9],
                    'weights': ['uniform', 'distance']
                }
            },
            'XGBoost': {
                'model': XGBClassifier(random_state=42),
                'params': {
                    'n_estimators': [100, 200],
                    'max_depth': [3, 5, 7],
                    'learning_rate': [0.01, 0.1]
                }
            }
        }
        self.best_models = {}
        self.feature_importances = {}
        self.scaler = StandardScaler()
        self.encoders = {}
        
    def prepare_data(self):
        """Prepare data with enhanced feature engineering"""
        print("Starting data preparation...")
        
        # Create target variable
        self.df['dropout'] = self.df['Status Description'].map(
            lambda x: 1 if x in ['Dropped Out', 'Withdraw'] else 0
        )
        
        # Convert dates and create features
        date_columns = ['Learner SignUp DateTime', 'Opportunity End Date', 
                        'Date of Birth', 'Graduation Date', 'Entry created at', 
                        'Apply Date', 'Opportunity Start Date']
        
        for col in date_columns:
            self.df[col] = pd.to_datetime(self.df[col], format='%d-%m-%Y', errors='coerce')
        
        # Enhanced feature engineering
        self.df['days_to_start'] = (self.df['Opportunity Start Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_end'] = (self.df['Opportunity End Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_graduate'] = (self.df['Graduation Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['application_processing_time'] = (self.df['Entry created at'] - self.df['Apply Date']).dt.days
        self.df['program_duration'] = (self.df['Opportunity End Date'] - self.df['Opportunity Start Date']).dt.days
        
        # Handle categorical variables
        categorical_columns = ['Gender', 'Country', 'Current Student Status', 
                               'Current/Intended Major', 'Opportunity Category']
        
        for col in categorical_columns:
            self.encoders[col] = LabelEncoder()
            self.df[col + '_encoded'] = self.encoders[col].fit_transform(self.df[col])
        
        # Select features
        self.feature_columns = [
            'Age', 'days_to_start', 'days_to_end', 'days_to_graduate',
            'application_processing_time', 'program_duration',
            'Gender_encoded', 'Country_encoded', 'Current Student Status_encoded',
            'Current/Intended Major_encoded', 'Opportunity Category_encoded'
        ]
        
        # Prepare X and y
        X = self.df[self.feature_columns]
        y = self.df['dropout']
        
        # Scale features
        X_scaled = self.scaler.fit_transform(X)
        X_scaled = pd.DataFrame(X_scaled, columns=self.feature_columns)
        
        # Split data with stratification
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            X_scaled, y, test_size=0.2, random_state=42, stratify=y
        )
        
        print("\nData preparation completed:")
        print(f"Training set shape: {self.X_train.shape}")
        print(f"Testing set shape: {self.X_test.shape}")
        print(f"Dropout rate: {self.y_train.mean():.2%}")
        
    def train_and_evaluate_models(self):
        """Train models with hyperparameter tuning and comprehensive evaluation"""
        results = []
        
        for name, model_info in self.models.items():
            print(f"\nTraining {name} with hyperparameter tuning...")
            
            # Perform grid search
            grid_search = GridSearchCV(
                model_info['model'],
                model_info['params'],
                cv=5,
                scoring='roc_auc',
                n_jobs=-1
            )
            grid_search.fit(self.X_train, self.y_train)
            
            # Store best model
            self.best_models[name] = grid_search.best_estimator_
            
            # Make predictions
            y_pred = grid_search.predict(self.X_test)
            y_pred_proba = grid_search.predict_proba(self.X_test)[:, 1]
            
            # Calculate metrics
            accuracy = grid_search.score(self.X_test, self.y_test)
            auc_score = roc_auc_score(self.y_test, y_pred_proba)
            cv_scores = cross_val_score(grid_search.best_estimator_, 
                                         self.X_train, self.y_train, cv=5)
            
            # Store results
            results.append({
                'Model': name,
                'Accuracy': accuracy,
                'AUC-ROC': auc_score,
                'CV Mean': cv_scores.mean(),
                'CV Std': cv_scores.std(),
                'Best Parameters': grid_search.best_params_
            })
            
            # Print detailed report
            print(f"\nBest parameters for {name}:")
            print(grid_search.best_params_)
            print(f"\nClassification Report for {name}:")
            print(classification_report(self.y_test, y_pred))
            
            # Calculate and plot confusion matrix
            self.plot_confusion_matrix(y_pred, name)
            
            # Calculate ROC curve
            self.plot_roc_curve(y_pred_proba, name)
            
            # Calculate feature importance
            if hasattr(grid_search.best_estimator_, 'feature_importances_'):
                self.feature_importances[name] = pd.DataFrame({
                    'feature': self.feature_columns,
                    'importance': grid_search.best_estimator_.feature_importances_
                }).sort_values('importance', ascending=False)
            elif name == 'Logistic Regression':
                self.feature_importances[name] = pd.DataFrame({
                    'feature': self.feature_columns,
                    'importance': abs(grid_search.best_estimator_.coef_[0])
                }).sort_values('importance', ascending=False)

        self.results_df = pd.DataFrame(results)
        self.plot_model_comparison()
        return self.results_df

    def plot_confusion_matrix(self, y_pred, model_name):
        """Plot confusion matrix for the model"""
        cm = confusion_matrix(self.y_test, y_pred)

        # Print the confusion matrix
        print(f"\nConfusion Matrix for {model_name}:\n")
        print(pd.DataFrame(cm, columns=['Predicted: Not Dropped Out', 'Predicted: Dropped Out'],
                           index=['True: Not Dropped Out', 'True: Dropped Out']))

        plt.figure(figsize=(8, 6))
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False,
                    xticklabels=['Not Dropped Out', 'Dropped Out'],
                    yticklabels=['Not Dropped Out', 'Dropped Out'])
        plt.xlabel('Predicted')
        plt.ylabel('True')
        plt.title(f'Confusion Matrix for {model_name}')
        plt.show()

    def plot_roc_curve(self, y_pred_proba, model_name):
        """Plot ROC curve for the model"""
        fpr, tpr, _ = roc_curve(self.y_test, y_pred_proba)
        plt.figure(figsize=(8, 6))
        plt.plot(fpr, tpr, label='ROC curve', color='blue')
        plt.plot([0, 1], [0, 1], color='red', linestyle='--')
        plt.xlabel('False Positive Rate')
        plt.ylabel('True Positive Rate')
        plt.title(f'ROC Curve for {model_name}')
        plt.legend()
        plt.show()

    def plot_model_comparison(self):
        """Visualize comparison of model performance metrics"""
        plt.figure(figsize=(12, 8))
        sns.barplot(data=self.results_df, x='Model', y='AUC-ROC', palette='viridis')
        plt.title('Model Comparison: AUC-ROC')
        plt.xticks(rotation=45)
        plt.show()

    def save_best_model(self, model_name):
        """Save the best Random Forest model to disk"""
        if model_name in self.best_models:
            joblib.dump(self.best_models[model_name], f'{model_name.replace(" ", "_")}_model.pkl')
            print(f"Saved {model_name} model to disk.")
        else:
            print(f"{model_name} not found in the best models.")

    def feature_importance_summary(self):
        for name, importance_df in self.feature_importances.items():
            plt.figure(figsize=(10, 6))
            sns.barplot(data=importance_df.head(10), x='importance', y='feature', palette='viridis')
            # Add values on bars
            for index, value in enumerate(importance_df.head(10)['importance']):
                plt.text(value, index, f'{value:.2f}', va='center')  # Adjust the format as needed
            plt.title(f'Feature Importance for {name}')
            plt.xlabel('Importance')
            plt.ylabel('Feature')
            plt.show()

# Example of using the class with a dataframe
df=pd.read_csv(r"C:\Users\vamsh\Downloads\Book1sample.csv")
predictor = AdvancedStudentDropoutPredictor(df)
predictor.prepare_data()
results = predictor.train_and_evaluate_models()
print(results)
predictor.save_best_model('Decision Tree')
predictor.feature_importance_summary()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
from collections import defaultdict

class StudentRecommendationSystem:
    def __init__(self, df):
        self.df = df.copy()
        self.encoders = {}
        self.scaler = StandardScaler()
        self.similarity_matrix = None
        self.feature_matrix = None
        self.student_profiles = None
        
    def prepare_data(self):
        """Prepare and engineer features for the recommendation system"""
        print("Preparing data for recommendation system...")
        
        # Convert dates to datetime
        date_columns = ['Learner SignUp DateTime', 'Opportunity End Date', 
                       'Date of Birth', 'Graduation Date', 'Entry created at', 
                       'Apply Date', 'Opportunity Start Date']
        
        for col in date_columns:
            self.df[col] = pd.to_datetime(self.df[col], format='%d-%m-%Y', errors='coerce')
            
        # Create engineered features
        self.df['days_to_start'] = (self.df['Opportunity Start Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_end'] = (self.df['Opportunity End Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_graduate'] = (self.df['Graduation Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['application_processing_time'] = (self.df['Entry created at'] - self.df['Apply Date']).dt.days
        
        # Encode categorical variables
        categorical_columns = ['Gender', 'Country', 'Current Student Status', 
                             'Current/Intended Major', 'Opportunity Category',
                             'Status Description']
        
        for col in categorical_columns:
            self.encoders[col] = LabelEncoder()
            self.df[col + '_encoded'] = self.encoders[col].fit_transform(self.df[col])
            
        # Select features for similarity calculation
        self.feature_columns = [
            'Age', 'days_to_start', 'days_to_end', 'days_to_graduate',
            'application_processing_time', 'Gender_encoded', 'Country_encoded',
            'Current Student Status_encoded', 'Current/Intended Major_encoded',
            'Opportunity Category_encoded'
        ]
        
        # Create feature matrix
        self.feature_matrix = self.df[self.feature_columns].copy()
        
        # Scale features
        self.feature_matrix = pd.DataFrame(
            self.scaler.fit_transform(self.feature_matrix),
            columns=self.feature_columns
        )
        
        # Calculate similarity matrix
        self.similarity_matrix = cosine_similarity(self.feature_matrix)
        
        # Create student profiles
        self.student_profiles = self.df[['Full_Name', 'Age', 'Gender', 'Current Student Status',
                                       'Current/Intended Major', 'Status Description']].copy()
        
        print("Data preparation completed.")
        
    def get_similar_students(self, student_name, n=5):
        """Find similar students based on feature similarity"""
        if student_name not in self.df['Full_Name'].values:
            raise ValueError(f"Student '{student_name}' not found in the dataset")
            
        student_idx = self.df[self.df['Full_Name'] == student_name].index[0]
        similar_indices = self.similarity_matrix[student_idx].argsort()[::-1][1:n+1]
        
        similar_students = self.student_profiles.iloc[similar_indices].copy()
        similar_students['similarity_score'] = self.similarity_matrix[student_idx][similar_indices]
        
        return similar_students
    
    def get_course_recommendations(self, student_name):
        """Recommend courses based on similar successful students"""
        similar_students = self.get_similar_students(student_name, n=5)
        
        # Filter successful students (not dropped out)
        successful_similar = similar_students[
            ~similar_students['Status Description'].isin(['Dropped Out', 'Withdraw'])
        ]
        
        # Get their courses
        successful_courses = self.df[
            self.df['Full_Name'].isin(successful_similar['Full_Name'])
        ]['Opportunity Name'].unique()
        
        # Get student's current courses
        student_courses = self.df[
            self.df['Full_Name'] == student_name
        ]['Opportunity Name'].unique()
        
        # Recommend new courses
        recommended_courses = [
            course for course in successful_courses 
            if course not in student_courses
        ]
        
        return recommended_courses
    
    def get_engagement_recommendations(self, student_name):
        """Generate personalized engagement recommendations"""
        student_data = self.df[self.df['Full_Name'] == student_name].iloc[0]
        recommendations = []
        
        # Check application processing time
        if student_data['application_processing_time'] > 7:
            recommendations.append(
                "Consider following up on your application status regularly"
            )
            
        # Check program duration
        if student_data['days_to_end'] - student_data['days_to_start'] > 180:
            recommendations.append(
                "This is a longer program - consider creating a study schedule"
            )
            
        # Check student status
        if student_data['Current Student Status'] == 'Graduate Program Student':
            recommendations.append(
                "Connect with other graduate students in similar programs"
            )
            
        # Add general recommendations
        recommendations.extend([
            "Participate in online discussion forums",
            "Set up regular check-ins with mentors",
            "Join study groups or form one"
        ])
        
        return recommendations
    
    def generate_comprehensive_recommendations(self, student_name):
        """Generate comprehensive recommendations for a student"""
        try:
            similar_students = self.get_similar_students(student_name)
            course_recommendations = self.get_course_recommendations(student_name)
            engagement_recommendations = self.get_engagement_recommendations(student_name)
            
            return {
                'similar_students': similar_students,
                'course_recommendations': course_recommendations,
                'engagement_recommendations': engagement_recommendations
            }
        except Exception as e:
            print(f"Error generating recommendations: {str(e)}")
            return None

def test_recommendation_system():
    """Test cases for the recommendation system"""
    # Sample test data
    test_data = pd.DataFrame({
        'Full_Name': ['Test Student 1', 'Test Student 2'],
        'Age': [24, 25],
        'Gender': ['Male', 'Female'],
        'Current Student Status': ['Graduate Program Student', 'Undergraduate Student'],
        'Current/Intended Major': ['Computer Science', 'Information Systems'],
        'Status Description': ['Started', 'Started'],
        'Learner SignUp DateTime': ['01-01-2024', '02-01-2024'],
        'Opportunity End Date': ['01-06-2024', '02-06-2024'],
        'Date of Birth': ['01-01-2000', '02-01-1999'],
        'Graduation Date': ['01-05-2024', '02-05-2024'],
        'Entry created at': ['05-01-2024', '06-01-2024'],
        'Apply Date': ['01-01-2024', '02-01-2024'],
        'Opportunity Start Date': ['15-01-2024', '16-01-2024'],
        'Opportunity Name': ['Test Course 1', 'Test Course 2'],
        'Opportunity Category': ['Course', 'Course'],
        'Country': ['United States', 'India']
    })
    
    # Initialize and test recommendation system
    rec_system = StudentRecommendationSystem(test_data)
    rec_system.prepare_data()
    
    # Test case 1: Get similar students
    similar_students = rec_system.get_similar_students('Test Student 1')
    assert len(similar_students) > 0, "Similar students not found"
    
    # Test case 2: Get course recommendations
    course_recs = rec_system.get_course_recommendations('Test Student 1')
    assert isinstance(course_recs, list), "Course recommendations not returned as list"
    
    # Test case 3: Get engagement recommendations
    engagement_recs = rec_system.get_engagement_recommendations('Test Student 1')
    assert len(engagement_recs) > 0, "No engagement recommendations generated"
    
    print("All test cases passed successfully!")

# Example usage
if __name__ == "__main__":
    # Load your data
    df=pd.read_csv(r"C:\Users\vamsh\Downloads\Book1sample.csv")
    
    # Initialize recommendation system
    rec_system = StudentRecommendationSystem(df)
    rec_system.prepare_data()
    
    # Generate recommendations for a specific student
    student_name = "Emmanuel Thompson"  # Example student
    recommendations = rec_system.generate_comprehensive_recommendations(student_name)
    
    # Print recommendations
    if recommendations:
        print(f"\nRecommendations for {student_name}:")
        print("\nSimilar Students:")
        print(recommendations['similar_students'])
        print("\nRecommended Courses:")
        for course in recommendations['course_recommendations']:
            print(f"- {course}")
        print("\nEngagement Recommendations:")
        for rec in recommendations['engagement_recommendations']:
            print(f"- {rec}")
            
    # Run test cases
    test_recommendation_system()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import TfidfVectorizer
import warnings
warnings.filterwarnings('ignore')

class StudentRecommendationSystem:
    def __init__(self):
        self.encoders = {}
        self.scaler = StandardScaler()
        self.similarity_matrix = None
        self.opportunity_similarity = None
        self.student_profiles = None
        self.opportunities = None
        
    def prepare_data(self, df):
        """
        Prepare and transform the data for the recommendation system
        """
        self.df = df.copy()
        
        # Convert dates to datetime
        date_columns = ['Learner SignUp DateTime', 'Opportunity End Date', 
                       'Date of Birth', 'Graduation Date', 'Entry created at', 
                       'Apply Date', 'Opportunity Start Date']
        
        for col in date_columns:
            self.df[col] = pd.to_datetime(self.df[col], format='%d-%m-%Y', errors='coerce')
            
        # Create engagement score (example metric)
        self.df['engagement_score'] = self.calculate_engagement_score()
        
        # Encode categorical variables
        categorical_columns = ['Gender', 'Country', 'Current Student Status', 
                             'Current/Intended Major', 'Opportunity Category']
        
        for col in categorical_columns:
            self.encoders[col] = LabelEncoder()
            self.df[col + '_encoded'] = self.encoders[col].fit_transform(self.df[col])
            
        # Create student profiles
        self.create_student_profiles()
        
        # Create opportunity profiles
        self.create_opportunity_profiles()
        
        return self.df
    
    def calculate_engagement_score(self):
        """
        Calculate student engagement score based on various factors
        """
        # Initialize base score
        base_score = 50
        
        # Adjust score based on status
        status_scores = {
            'Started': 20,
            'Team Allocated': 15,
            'Waitlisted': 5,
            'Rejected': -10,
            'Dropped Out': -20,
            'Withdraw': -15
        }
        
        engagement_scores = self.df['Status Description'].map(status_scores).fillna(0) + base_score
        
        # Adjust score based on application timing
        application_delay = (self.df['Apply Date'] - self.df['Learner SignUp DateTime']).dt.days
        engagement_scores -= np.clip(application_delay * 0.1, 0, 10)  # Penalty for delayed application
        
        return engagement_scores
    
    def create_student_profiles(self):
        """
        Create student profiles based on their characteristics and behavior
        """
        profile_features = [
            'Age', 'Gender_encoded', 'Country_encoded', 
            'Current Student Status_encoded', 'Current/Intended Major_encoded',
            'engagement_score'
        ]
        
        self.student_profiles = self.df[profile_features].copy()
        self.student_profiles = pd.DataFrame(
            self.scaler.fit_transform(self.student_profiles),
            columns=profile_features,
            index=self.df.index
        )
        
        # Calculate similarity matrix for collaborative filtering
        self.similarity_matrix = cosine_similarity(self.student_profiles)
        
    def create_opportunity_profiles(self):
        """
        Create opportunity profiles for content-based filtering
        """
        # Combine relevant features into a text representation
        self.df['opportunity_features'] = (
            self.df['Opportunity Category'] + ' ' +
            self.df['Current/Intended Major']
        )
        
        # Create TF-IDF matrix
        tfidf = TfidfVectorizer(stop_words='english')
        opportunity_tfidf = tfidf.fit_transform(self.df['opportunity_features'])
        
        # Calculate opportunity similarity
        self.opportunity_similarity = cosine_similarity(opportunity_tfidf)
        
    def get_similar_students(self, student_index, n_similar=5):
        """
        Find similar students based on profiles
        """
        student_similarities = self.similarity_matrix[student_index]
        similar_indices = student_similarities.argsort()[::-1][1:n_similar+1]
        
        similar_students = pd.DataFrame({
            'Student': self.df.iloc[similar_indices]['Full_Name'],
            'Similarity Score': student_similarities[similar_indices],
            'Major': self.df.iloc[similar_indices]['Current/Intended Major'],
            'Status': self.df.iloc[similar_indices]['Status Description']
        })
        
        return similar_students
    
    def get_opportunity_recommendations(self, student_index, n_recommendations=5):
        """
        Get opportunity recommendations based on content similarity
        """
        # Get opportunities similar to the student's current opportunity
        opportunity_scores = self.opportunity_similarity[student_index]
        
        # Exclude current opportunity and sort by similarity
        opportunity_scores[student_index] = -1
        top_opportunities = opportunity_scores.argsort()[::-1][:n_recommendations]
        
        recommendations = pd.DataFrame({
            'Opportunity Name': self.df.iloc[top_opportunities]['Opportunity Name'],
            'Category': self.df.iloc[top_opportunities]['Opportunity Category'],
            'Similarity Score': opportunity_scores[top_opportunities]
        })
        
        return recommendations
    
    def get_personalized_recommendations(self, student_index, n_recommendations=5):
        """
        Get personalized recommendations combining collaborative and content-based filtering
        """
        # Get similar students
        similar_students = self.get_similar_students(student_index, n_similar=3)
        similar_indices = similar_students.index
        
        # Get opportunities taken by similar students
        similar_opportunities = set()
        for idx in similar_indices:
            similar_opportunities.update(
                self.df[self.df.index != student_index]['Opportunity Name'].unique()
            )
        
        # Score opportunities based on content similarity
        opportunity_scores = {}
        student_opportunity = self.df.iloc[student_index]['Opportunity Name']
        
        for opportunity in similar_opportunities:
            if opportunity != student_opportunity:
                opp_idx = self.df[self.df['Opportunity Name'] == opportunity].index[0]
                content_score = self.opportunity_similarity[student_index][opp_idx]
                collaborative_score = np.mean([
                    self.similarity_matrix[student_index][idx] 
                    for idx in similar_indices
                ])
                # Combine scores (weighted average)
                final_score = 0.7 * content_score + 0.3 * collaborative_score
                opportunity_scores[opportunity] = final_score
        
        # Get top recommendations
        top_opportunities = sorted(
            opportunity_scores.items(), 
            key=lambda x: x[1], 
            reverse=True
        )[:n_recommendations]
        
        recommendations = pd.DataFrame(
            top_opportunities, 
            columns=['Opportunity Name', 'Score']
        )
        
        return recommendations
    
    def get_risk_based_recommendations(self, student_index):
        """
        Get recommendations based on student's risk profile
        """
        student_data = self.df.iloc[student_index]
        engagement_score = student_data['engagement_score']
        
        recommendations = {
            'risk_level': '',
            'suggestions': []
        }
        
        # Determine risk level
        if engagement_score < 40:
            recommendations['risk_level'] = 'High Risk'
            recommendations['suggestions'] = [
                "Schedule immediate academic counseling session",
                "Connect with peer mentor",
                "Join study groups for current courses",
                "Consider shorter duration opportunities",
                "Access additional learning resources"
            ]
        elif engagement_score < 60:
            recommendations['risk_level'] = 'Medium Risk'
            recommendations['suggestions'] = [
                "Regular check-ins with academic advisor",
                "Participate in skill-building workshops",
                "Join relevant student communities",
                "Explore additional learning opportunities"
            ]
        else:
            recommendations['risk_level'] = 'Low Risk'
            recommendations['suggestions'] = [
                "Continue current engagement level",
                "Consider mentoring other students",
                "Explore advanced opportunities",
                "Share success strategies with peers"
            ]
        
        return recommendations

def test_recommendation_system():
    """
    Test cases for the recommendation system
    """
    # Sample test data
    test_data = pd.DataFrame({
        'Learner SignUp DateTime': ['14-06-2023', '01-05-2023', '09-04-2023'],
        'Opportunity Name': ['Course A', 'Course B', 'Course C'],
        'Opportunity Category': ['Course', 'Competition', 'Internship'],
        'Opportunity End Date': ['29-06-2024', '29-06-2024', '29-06-2024'],
        'Date of Birth': ['12-01-2001', '16-08-2000', '27-01-2002'],
        'Gender': ['Female', 'Female', 'Male'],
        'Country': ['USA', 'India', 'USA'],
        'Current Student Status': ['Undergraduate', 'Graduate', 'Undergraduate'],
        'Current/Intended Major': ['Computer Science', 'Data Science', 'Engineering'],
        'Status Description': ['Started', 'Started', 'Dropped Out'],
        'Apply Date': ['14-06-2023', '01-05-2023', '09-04-2023'],
        'Full_Name': ['Student A', 'Student B', 'Student C']
    })
    
    # Initialize and test recommendation system
    print("Testing Recommendation System...")
    
    try:
        # Initialize system
        rec_system = StudentRecommendationSystem()
        rec_system.prepare_data(test_data)
        
        # Test 1: Similar Students
        print("\nTest 1: Getting Similar Students")
        similar_students = rec_system.get_similar_students(0)
        print(similar_students)
        
        # Test 2: Opportunity Recommendations
        print("\nTest 2: Getting Opportunity Recommendations")
        opp_recommendations = rec_system.get_opportunity_recommendations(0)
        print(opp_recommendations)
        
        # Test 3: Personalized Recommendations
        print("\nTest 3: Getting Personalized Recommendations")
        personalized_recs = rec_system.get_personalized_recommendations(0)
        print(personalized_recs)
        
        # Test 4: Risk-Based Recommendations
        print("\nTest 4: Getting Risk-Based Recommendations")
        risk_recs = rec_system.get_risk_based_recommendations(0)
        print(f"Risk Level: {risk_recs['risk_level']}")
        print("Suggestions:")
        for suggestion in risk_recs['suggestions']:
            print(f"- {suggestion}")
        
        print("\nAll tests completed successfully!")
        
    except Exception as e:
        print(f"Error during testing: {str(e)}")

# Example usage with your actual dataset
if __name__ == "__main__":
    # First run test cases
    test_recommendation_system()
    
    # Then use with actual data
    try:
        # Load your actual dataset
        df=pd.read_csv(r"C:\Users\vamsh\Downloads\Book1sample.csv")
        
        # Initialize recommendation system
        rec_system = StudentRecommendationSystem()
        processed_data = rec_system.prepare_data(df)
        
        # Example: Get recommendations for first student
        student_index = 0
        
        print("\nRecommendations for student:", df.iloc[student_index]['Full_Name'])
        
        # Get similar students
        similar_students = rec_system.get_similar_students(student_index)
        print("\nSimilar Students:")
        print(similar_students)
        
        # Get opportunity recommendations
        opp_recommendations = rec_system.get_opportunity_recommendations(student_index)
        print("\nRecommended Opportunities:")
        print(opp_recommendations)
        
        # Get personalized recommendations
        personalized_recs = rec_system.get_personalized_recommendations(student_index)
        print("\nPersonalized Recommendations:")
        print(personalized_recs)
        
        # Get risk-based recommendations
        risk_recs = rec_system.get_risk_based_recommendations(student_index)
        print("\nRisk Assessment and Recommendations:")
        print(f"Risk Level: {risk_recs['risk_level']}")
        print("Suggested Actions:")
        for suggestion in risk_recs['suggestions']:
            print(f"- {suggestion}")
            
    except Exception as e:
        print(f"Error processing actual dataset: {str(e)}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
import joblib
from datetime import datetime

class StudentDropoutRiskRecommender:
    def __init__(self, df):
        self.df = df.copy()
        self.encoders = {}
        self.scaler = StandardScaler()
        self.model = RandomForestClassifier(
            n_estimators=200,
            max_depth=10,
            min_samples_split=5,
            random_state=42
        )
        self.risk_thresholds = {
            'low': 0.3,
            'medium': 0.6,
            'high': 0.8
        }
        
    def prepare_data(self):
        """Prepare data for dropout prediction"""
        print("Preparing data for dropout risk assessment...")
        
        # Create target variable
        self.df['dropout'] = self.df['Status Description'].map(
            lambda x: 1 if x in ['Dropped Out', 'Withdraw'] else 0
        )
        
        # Convert dates to datetime
        date_columns = [
            'Learner SignUp DateTime', 'Opportunity End Date', 
            'Date of Birth', 'Graduation Date', 'Entry created at', 
            'Apply Date', 'Opportunity Start Date'
        ]
        
        for col in date_columns:
            self.df[col] = pd.to_datetime(self.df[col], format='%d-%m-%Y', errors='coerce')
        
        # Create engineered features
        self.df['days_to_start'] = (self.df['Opportunity Start Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_end'] = (self.df['Opportunity End Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['days_to_graduate'] = (self.df['Graduation Date'] - self.df['Learner SignUp DateTime']).dt.days
        self.df['application_processing_time'] = (self.df['Entry created at'] - self.df['Apply Date']).dt.days
        self.df['program_duration'] = (self.df['Opportunity End Date'] - self.df['Opportunity Start Date']).dt.days
        
        # Encode categorical variables
        categorical_columns = [
            'Gender', 'Country', 'Current Student Status', 
            'Current/Intended Major', 'Opportunity Category'
        ]
        
        for col in categorical_columns:
            self.encoders[col] = LabelEncoder()
            self.df[col + '_encoded'] = self.encoders[col].fit_transform(self.df[col])
        
        # Select features for model
        self.feature_columns = [
            'Age', 'days_to_start', 'days_to_end', 'days_to_graduate',
            'application_processing_time', 'program_duration',
            'Gender_encoded', 'Country_encoded', 'Current Student Status_encoded',
            'Current/Intended Major_encoded', 'Opportunity Category_encoded'
        ]
        
        # Prepare features and target
        X = self.df[self.feature_columns]
        y = self.df['dropout']
        
        # Scale features
        X_scaled = self.scaler.fit_transform(X)
        
        # Split data
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            X_scaled, y, test_size=0.2, random_state=42, stratify=y
        )
        
        print("Data preparation completed.")
        
    def train_model(self):
        """Train the dropout prediction model"""
        print("Training dropout prediction model...")
        self.model.fit(self.X_train, self.y_train)
        
        # Evaluate model
        y_pred = self.model.predict(self.X_test)
        print("\nModel Performance:")
        print(classification_report(self.y_test, y_pred))
        
        # Get feature importance
        self.feature_importance = pd.DataFrame({
            'feature': self.feature_columns,
            'importance': self.model.feature_importances_
        }).sort_values('importance', ascending=False)
        
        print("\nTop 5 Risk Factors:")
        print(self.feature_importance.head())
        
    def get_risk_profile(self, student_data):
        """Generate risk profile for a student"""
        # Prepare student features
        student_features = student_data[self.feature_columns].values.reshape(1, -1)
        student_features_scaled = self.scaler.transform(student_features)
        
        # Get dropout probability
        dropout_prob = self.model.predict_proba(student_features_scaled)[0][1]
        
        # Determine risk level
        if dropout_prob < self.risk_thresholds['low']:
            risk_level = 'Low'
        elif dropout_prob < self.risk_thresholds['medium']:
            risk_level = 'Medium'
        elif dropout_prob < self.risk_thresholds['high']:
            risk_level = 'High'
        else:
            risk_level = 'Very High'
            
        return {
            'dropout_probability': dropout_prob,
            'risk_level': risk_level
        }
        
    def get_risk_factors(self, student_data):
        """Identify specific risk factors for a student"""
        risk_factors = []
        
        # Check application processing time
        if student_data['application_processing_time'].iloc[0] > 7:
            risk_factors.append({
                'factor': 'Long application processing time',
                'recommendation': 'Follow up on your application status and ensure all required documents are submitted'
            })
            
        # Check program duration
        if student_data['program_duration'].iloc[0] > 180:
            risk_factors.append({
                'factor': 'Long program duration',
                'recommendation': 'Create a detailed study schedule and set regular milestones'
            })
            
        # Check days to start
        if student_data['days_to_start'].iloc[0] < 0:
            risk_factors.append({
                'factor': 'Late program start',
                'recommendation': 'Contact program coordinator for catch-up materials and extra support'
            })
            
        # Add major-specific recommendations
        major = student_data['Current/Intended Major'].iloc[0]
        if 'Computer' in major or 'Information' in major:
            risk_factors.append({
                'factor': 'Technical program complexity',
                'recommendation': 'Join study groups and utilize online learning resources'
            })
            
        return risk_factors
        
    def generate_recommendations(self, student_name):
        """Generate comprehensive risk assessment and recommendations"""
        if student_name not in self.df['Full_Name'].values:
            raise ValueError(f"Student '{student_name}' not found in the dataset")
            
        student_data = self.df[self.df['Full_Name'] == student_name]
        
        # Get risk profile
        risk_profile = self.get_risk_profile(student_data)
        
        # Get risk factors and recommendations
        risk_factors = self.get_risk_factors(student_data)
        
        # Generate intervention strategies based on risk level
        if risk_profile['risk_level'] in ['High', 'Very High']:
            interventions = [
                "Schedule immediate academic counseling session",
                "Assign a peer mentor",
                "Provide access to additional learning resources",
                "Set up weekly progress monitoring"
            ]
        else:
            interventions = [
                "Regular check-ins with academic advisor",
                "Join study groups or discussion forums",
                "Utilize available learning resources"
            ]
            
        return {
            'student_name': student_name,
            'risk_profile': risk_profile,
            'risk_factors': risk_factors,
            'recommended_interventions': interventions
        }
    
    def save_model(self, filename='dropout_risk_model.pkl'):
        """Save the trained model"""
        model_data = {
            'model': self.model,
            'scaler': self.scaler,
            'encoders': self.encoders,
            'feature_columns': self.feature_columns
        }
        joblib.dump(model_data, filename)
        print(f"Model saved as {filename}")

def test_risk_assessment_system():
    """Test cases for the risk assessment system"""
    # Create test data
    test_data = pd.DataFrame({
        'Full_Name': ['Faria Khan', 'Rachel Grant'],
        'Age': [23, 15],
        'Gender': ['Female', 'Female'],
        'Current Student Status': ['Undergraduate Student', 'High School Student'],
        'Current/Intended Major': ['Radiology', 'Computer Science and Information Systems'],
        'Status Description': ['Started', 'Dropped Out'],
        'Learner SignUp DateTime': ['14-06-2023', '07-02-2023'],
        'Opportunity End Date': ['29-06-2024', '28-02-2009'],
        'Date of Birth': ['12-01-2001', '02-01-1999'],
        'Graduation Date': ['01-11-2023', '01-01-2027'],
        'Entry created at': ['11-03-2024', '11-03-2024'],
        'Apply Date': ['14-06-2023', '07-02-2023'],
        'Opportunity Start Date': ['03-11-2022', '18-11-2021'],
        'Opportunity Name': ['Career Essentials: Getting Started with Your Professional', 'Slide Geeks: A Presentation Design Competition'],
        'Opportunity Category': ['Course', 'Competition'],
        'Country': ['United States', 'Kenya']
    })
    
    df = df=pd.read_csv(r"C:\Users\vamsh\Downloads\Book1sample.csv")
    recommender = StudentDropoutRiskRecommender(df)
    recommender.prepare_data()
    recommender.train_model()
    
    # Assess a test student
    student_name = "Faria Khan"
    recommendations = recommender.generate_recommendations(student_name)
    print("\nRecommendations for", student_name)
    print(recommendations)

# Run the test case
test_risk_assessment_system()
```
